## Disease Background: 

**Prevalence & incidence**: 30,000 to 40,000 new U.S. cases per year. Around 100,000 Americans live with IPF as of 2022. Death rates (age-adjusted) increased from 5.3 to 7.1 per 100,000 between 2017 and 2022. 70% of cases are in males. Onset is usually within 50-70 age range, with a median age at diagnosis of 62.

**Economic Burden**: Around $2 billion/year in U.S. medical costs (Medicare alone, excludes drugs). Per patient, total medical costs for a person with IPF are about $20,000 per year above matched controls without IPF.

**Risk factors**
- The strongest genetic single risk factor is the  MUC5B rs35705950 promoter variant. There is a 6–20× risk increase for IPF if person carries one or two copies of the risk T allele. Telomere-gene mutations such as TERT/TERC which result in short telomeres also pose a genetic risk factor.
- The primary lifestyle risk factor is cigarette smoking which doubles IPF risk. Occupational exposures including silica dust, wood dust, metal dusts, asbestos, and agricultural exposures pose a risk. The CDC reported in 2024 that around 21% or IPF deaths could be attributed to occuptational exposures.

**Societal Determinants**: Data that was presented at CHEST 2024 indicates that patients in the lowest income quartile had higher rates of pneumonia-triggered acute exacerbations compared to the highest quartile patients (29.4% vs. 24.9%). Black and Hispanic patients are less likely to receive antifibrotic therapy, and they may also have later-stage disease at diagnosis. This is due to reduced access to specialist centers, insurance gaps, and lower health literacy.

**Symptoms**: Crackling sound in the lungs, difficulty breathing, fatigue, and persistent dry cough. Symptom onset usually preceds diagnosis by one to two years. As IPF advances, shortness of breath occurs at rest, and patients have lower tolerance for exercise. They may experience oxygen desaturation during physical exertion. Digital clubbing is present in 25%-50% of patients. Patients may have fatigue, weight loss, and impaired quality of life. When the disease is advanced, pulmonary hypertension (strain on the right heart) can develop.

**Diagnosis**: High-resolution computed tomography (HRCT) is a specialized CT scan that obtains detailed images of the lungs. Honeycombing, subpleural cysts from collapsed fibrotic alveoli, is the hallmark radiologic feature of usual interstitial pneumonia (UIP) needed for IPF diagnosis. Honeycombing has a heterogeneous distribution with density typically increasing toward the bases of the lungs.

**Standard of care treatment(s)**: As of 2025, there are three FDA-approved treatments for IPF:
- Pirfenidone is an oral antifibrotic that works through downregulation of the production of growth factors and procollagen I and II.
- Nintedanib is an oral tyrosine kinase inhibitor. It binds to growth factor receptors which inhibits fibroblast proliferation.
- Nerandomilast is a phosphodiesterase 4 (PDE4) inhibitor. It works by interfering with the pathway that creates scar tissue (fibrosis) and inflammation in the lung tissue.
Non-pharmacologic management of IPF includes supplemental oxygen, pulmonary rehabilitation, influenza and pneumococcal vaccination, and hospice care for end-stage disease. A lung transplant is the only treatment that can improve survival in eligible patients. 986 IPF patients received a lung transplant in the U.S. in 2023.

**Disease Progression & Prognosis**: Disease progression varies: it can be slowly progressive, rapidly progressive, or episodic. Forced vital capacity (FVC) is a predictor of mortality if it declines by at least 10% over 12 months for the patient. Resting oxygen desaturation, degree of honeycombing on CT, and short telomere length are other prognostic indicators. Acute exacerbations occur in 5%-15% of IPF patients annually, and mortality rates for 30 days from acute exacerbations are 40%. The number of exacerbations directly increase mortality risk, while antifibrotic therapy can be an option to reduce their frequency. Common causes of death include respiratory failure, lung cancer, cardiovascular events, and infection.

**Continuum of Care Providers**: Primary care physicians are the first point of contact. Pulmonologists have a role in HRCT interpretation and diagnostic workup. They initiate and manage antifibrotic therapy and monitor disease progression. Radiologists interpret HRCT and classify UIP pattern in the patient, quantifying honeycombing. Lung transplant teams are involved in some cases. Respiratory therapists can administer oxygen therapy and pulmonary rehabilitation.

- Cellular Mechanisms: myofibroblasts overproliferate and produce excessive collagen, myofibroblasts don't undergo apoptosis (cell death) properly and they accumulate

**Clinical Trials & Next-Generation Therapies**

### Image Analysis Completed
The analysis stage processes 6 selected lung tissue images:
- Loads each image as grayscale using OpenCV.
- Applies binary thresholding (threshold = 127) to classify pixels as white (fibrotic, value 255) or black (healthy, value 0).
- Counts white and black pixels for each image.
- Calculates the percentage of white pixels as a proxy for fibrosis severity.
- Prints detailed results including pixel counts and percentages with corresponding depths.
- Saves the results to a CSV file ('Percent_White_Pixels.csv') for further analysis.
- Measures and reports total execution time.

```python
# Configuration

# choosing 6 images to analyze
FILENAMES = [
    r"images/MASK_SK658 Slobe ch010118.jpg",
    r"images/MASK_SK658 Slobe ch010113.jpg",        
    r"images/MASK_SK658 Slobe ch010098.jpg",
    r"images/MASK_SK658 Llobe ch010022.jpg",
    r"images/MASK_SK658 Slobe ch010089.jpg",
    r"images/MASK_SK658 Slobe ch010156.jpg"
]

# Load depths from CSV given with the image names and depths
depth_df = pd.read_csv("Filenames and Depths for Students.csv")
depth_dict = dict(zip(depth_df["Filenames"], depth_df["Depth from lung surface (in micrometers) where image was acquired"])) # this creates a dictionary that has each image filename as the key and then its depth is the value
DEPTHS = [depth_dict[fn] for fn in FILENAMES] # creates a list of depths that go in the order of the filenames
 
OUTPUT_CSV = "Percent_White_Pixels.csv" # Output CSV filename
THRESHOLD = 127 # this is the threshold value to create grayscale images on a binary scale. 
 
# this function loads the images and makes sure it's loaded correctly
def load_image(filepath: str) -> np.ndarray: 
    """Load a grayscale image; exit with a clear message if not found."""
    img = cv2.imread(filepath, cv2.IMREAD_GRAYSCALE) # Load as grayscale image
    if img is None: # Check if the image was loaded successfully
        print(colored(f"ERROR: Could not load image: {filepath}", "red")) # Print error message if image is not able to be loaded
        sys.exit(1) 
    return img # returns the image
 
# this function counts the number of white and black pixels in the image
def count_pixels(img: np.ndarray, threshold: int = THRESHOLD) -> tuple[int, int]: 
    """Return (white_count, black_count) for a grayscale image."""
    _, binary = cv2.threshold(img, threshold, 255, cv2.THRESH_BINARY) # Binarize the image using the specified threshold
    white = int(np.sum(binary == 255)) # Count white pixels (255 in binary image)
    black = int(np.sum(binary == 0)) # Count black pixels (0 in binary image)
    return white, black # Return the counts of white and black pixels as integers
 
# this function calculates the percentage of white pixels in the image
def white_percentage(white: int, black: int) -> float:
    total = white + black # Calculate total pixels
    return 100.0 * white / total if total > 0 else 0.0 # Return percentage of white pixels; handle division by zero
 
#this function analyszes the images in the filenames list and returns the counts of white and black pixels and the percentage of white pixels for each image
def analyze_images(filenames: list[str]) -> tuple[list[int], list[int], list[float]]: # type: (list[str]) -> tuple[list[int], list[int], list[float]]
    white_counts, black_counts, white_percents = [], [], [] # Initialize lists to store results
    for filepath in filenames: # Loop through each file path in the list of filenames
        img = load_image(filepath) # use the load_image function to load the image from the file path
        white, black = count_pixels(img) # use the count_pixels function to count the number of white and black pixels in the image
        "lines below add the pixel counts and percentage to the lists for each image"
        white_counts.append(white)
        black_counts.append(black)
        white_percents.append(white_percentage(white, black))
    return white_counts, black_counts, white_percents # Return the lists of white counts, black counts, and white percentages for all images


#this function prints the counts of white and black pixels for each image in the list of filenames
def print_pixel_counts(filenames, white_counts, black_counts): 
    print(colored("\nCounts of pixels by color in each image", "yellow"))
    for i, filepath in enumerate(filenames): # Loop through each file path and corresponding pixel counts
        print(colored(f"  Image {i} — White: {white_counts[i]:,}  |  Black: {black_counts[i]:,}", "white")) # Print pixel counts for each image in white
 
# this function prints the percentage of white pixels and corresponding depths for each image in the list of filenames
def print_white_percents(filenames, depths, white_percents): 
    print(colored("\nPercent white pixels per image", "yellow")) 
    for i, filepath in enumerate(filenames): # Loop through each file path and corresponding white pixel percentage
        print(colored(f"  {filepath}", "red"))
        print(f"    {white_percents[i]:.2f}% white  |  Depth: {depths[i]:,} µm\n") # Print the percentage of white pixels and corresponding depth for each image

# this function saves the results of the analysis to a CSV file
def save_csv(filenames, depths, white_percents, output_path: str = OUTPUT_CSV): 
    df = pd.DataFrame({ # Create a DataFrame to store results
        "Filename":      filenames,
        "Depth (µm)":   depths,
        "White (%)":    [round(p, 4) for p in white_percents],
    })
    df.to_csv(output_path, index=False) # Save the DataFrame to a CSV file without the index
    print(colored(f"Results saved to '{output_path}'.", "green"))

if __name__ == "__main__": # Run the analysis and reporting when the script is executed
    white_counts, black_counts, white_percents = analyze_images(FILENAMES) # Analyze the images and get pixel counts and percentages
    print_pixel_counts(FILENAMES, white_counts, black_counts) # Print the pixel counts for each image
    print_white_percents(FILENAMES, DEPTHS, white_percents) # Print the percentage of white pixels and corresponding depths for each image
    save_csv(FILENAMES, DEPTHS, white_percents) # Save the results to a CSV file

if __name__ == "__main__": # Run the analysis and reporting when the script is executed, and also measure runtime
    start = time.time() # Record the start time

    white_counts, black_counts, white_percents = analyze_images(FILENAMES) # Analyze the images and get pixel counts and percentages
    print_pixel_counts(FILENAMES, white_counts, black_counts) # Print the pixel counts for each image
    print_white_percents(FILENAMES, DEPTHS, white_percents) # Print the percentage of white pixels and corresponding depths for each image
    save_csv(FILENAMES, DEPTHS, white_percents) # Save the results to a CSV file

    end = time.time() # Record the end time
    print(f"\nTotal runtime: {end - start:.4f} seconds") # Print the total runtime of the analysis and reporting in seconds
```

### Interpolation Methods Overview
This stage performs predictive modeling using interpolation:
- Prepares data by handling duplicate depths (averaging percentages) and sorting.
- Uses scipy's `interp1d` function with three methods: linear (straight lines), quadratic (parabolic curves), and cubic (smoother curves).
- Interpolates fibrosis percentages at depths 1500, 3000, 4500, 6000, 7500, and 9000 µm.
- Allows extrapolation beyond the measured data range.
- Compares methods to assess sensitivity of predictions to interpolation technique.

```python
if __name__ == "__main__":
    # Record the start time to measure total execution time
    start_time = time.time()


    # Record end time and print runtime
    end_time = time.time()
    print(f"\nTotal runtime: {end_time - start_time:.4f} seconds")

    # Interpolation Section

    # Prepare data for interpolation by handling potential duplicate depths
    # First, pair depths with their corresponding white percentages
    data = list(zip(DEPTHS, white_percents))
    # Sort the data by depth (first element of each pair)
    data.sort(key=lambda x: x[0])

    # Initialize lists for unique depths and averaged percentages
    unique_depths = []
    avg_percents = []

    # Use defaultdict to group percentages by depth
    from collections import defaultdict
    depth_to_percents = defaultdict(list)
    # Populate the dictionary: key is depth, value is list of percentages at that depth
    for d, p in data:
        depth_to_percents[d].append(p)

    # For each unique depth, calculate the average percentage (handles duplicates)
    for d in sorted(depth_to_percents.keys()):
        unique_depths.append(d)
        avg_percents.append(np.mean(depth_to_percents[d]))

    # Convert to numpy arrays for efficient numerical operations
    x_sorted = np.array(unique_depths)  # Sorted unique depths
    y_sorted = np.array(avg_percents)   # Corresponding averaged percentages

    # Define the depths at which we want to interpolate fibrosis percentages
    # These are depths where no images were taken, so we estimate values
    interp_depths = [1500, 3000, 4500, 6000, 7500, 9000]

    # List of interpolation methods to compare: linear, quadratic, and cubic
    methods = ['linear', 'quadratic', 'cubic']

    # Dictionary to store interpolation results: method -> list of (depth, interpolated_value) tuples
    interp_results = {method: [] for method in methods}

    # Perform interpolation for each depth and each method
    for depth in interp_depths:
        for method in methods:
            # Create interpolation function based on method
            if method == 'linear':
                # Linear interpolation: connects points with straight lines
                interp_fn = interp1d(x_sorted, y_sorted, kind='linear', fill_value="extrapolate")
            elif method == 'quadratic':
                # Quadratic interpolation: fits a quadratic curve (parabola) between points
                # Requires at least 3 data points
                if len(x_sorted) >= 3:
                    interp_fn = interp1d(x_sorted, y_sorted, kind='quadratic', fill_value="extrapolate")
                else:
                    print(f"Not enough points for {method}, skipping")
                    continue
            elif method == 'cubic':
                # Cubic interpolation: fits a cubic curve, smoother than quadratic
                # Requires at least 4 data points
                if len(x_sorted) >= 4:
                    interp_fn = interp1d(x_sorted, y_sorted, kind='cubic', fill_value="extrapolate")
                else:
                    print(f"Not enough points for {method}, skipping")
                    continue

            # Evaluate the interpolation function at the specified depth
            interpolated_y = float(interp_fn(depth))

            # Store the result
            interp_results[method].append((depth, interpolated_y))

            # Print the interpolated value with colored output
            print(colored(
                f"{method.capitalize()} interpolation at {depth} µm: {interpolated_y:.4f}% white pixels",
                "green"
            ))

    # Create a grid of subplots: rows = depths, columns = methods
    fig, axs = plt.subplots(len(interp_depths), len(methods), figsize=(15, 20))

    # Plot each interpolation result
    for i, depth in enumerate(interp_depths):
        for j, method in enumerate(methods):
            # Check if we have results for this method (in case it was skipped)
            if len(interp_results[method]) > i:
                ax = axs[i, j]
                # Plot the original data points as blue line with circles
                ax.plot(x_sorted, y_sorted, 'o-', color='blue', label='Data')
                # Plot the interpolated point as a red dot
                _, interp_y = interp_results[method][i]
                ax.scatter(depth, interp_y, color='red', s=100, label=f'Interpolated at {depth} µm')
                # Set plot title, labels, and grid
                ax.set_title(f'{method.capitalize()} at {depth} µm')
                ax.set_xlabel('Depth (µm)')
                ax.set_ylabel('% White Pixels')
                ax.grid(True)
                ax.legend()
            else:
                # Hide subplot if no data (method was skipped)
                axs[i, j].set_visible(False)

    # Adjust layout and display the plots
    plt.tight_layout()
    plt.show()
```

### Validation Comparison: Interpolated vs. Actual Measurements

To validate the interpolation methods, we analyzed images from the given dataset taken exactly at the interpolated depths (1500, 3000, 4500, 6000, 7500, 9000 µm). The actual measured fibrosis percentages are compared below to the interpolated predictions from the original dataset.

- **1500 µm**: Actual = 1.49%, Linear = 1.2353%, Quadratic = 2.1681%, Cubic = 3.3892%  
- **3000 µm**: Actual = 1.92%, Linear = 1.6553%, Quadratic = 5.5003%, Cubic = 8.8317%

- **9000 µm**: Actual = 3.84%, Linear = 4.6889%, Quadratic = -9.7181%, Cubic = -9.2462%  

The actual fibrosis percentage increases with depth, validating the biological expectation of deeper fibrosis accumulation. Linear interpolation closely follows the actual trend and is consistently within a sensible range, while quadratic and cubic diverge dramatically at deeper depths (overprediction early and unrealistic negative extrapolation at 9000 µm), demonstrating overfitting and instability in this sparse dataset.

With our limited depth data, **linear interpolation** is the recommended method. It balances simplicity and reliability, and it avoids the extreme overshoots in quadratic/cubic for this dataset. The dataset would need more intermediate-depth samples to safely apply higher-order interpolation for better curve fitting.


Based on our image analysis, we found that as you increase the depth the mouse biopsies are done in, the percentage of white pixels increases. These white pixels represent the percentage of fibrotic tissue found in the images of lung tissue. Prior literature supports this idea that increasing the depth also increases the amount of fibrotic tissue. One study, "The normal and fibrotic mouse lung classified by spatial proteomic analysis" also studied mice with bleomycin induced lung fibrosis. Researchers here found that how fibrotic tissue changes over time (Ciccimarra et al. 2022). As they conduct a single cell analysis on these mice, they found that density increased further into the lung tissue which aligns with the stiffened tissue that happens in pulmonary fibrosis. Additionally the paper explains how their   cell analysis results matched the idea that scar tissue formation rises over time. Though the paper uses spatial proteomic analysis as opposed what we did in the project with image slides,we were still able to observe the same pattern.  

## Conclusions and Ethical Implications:

Our project seeked to understand how the percentage of fibrotic tissue changed at different biopsy depths by using image analysis and use interpolation to predict these fibrotic levels. Based on our initial image analysis of 6 images, we found that as the depth of biopsy images increased, so did the amount of fibrotic tissue. From a depth of 80 µm to 1000 µm, the percentage went up from 0.79% to 8.82%, showing a stark increase. We also conducted different interpolation methods including linear, quadratic and cubic to see how these methods perfomed when predicting the biopsy depths. Visually on our interpolation graphs of the 6 images, it's clear that even when interpolation was conducted at different depths, the linear method was accurate throughout whereas the quadric and cubic methods were off. To further display the effectiveness of the linear method, we found images in the dataset that were taken at the depths we conducted interpolation at (1500µm to 9000µm in 1500 increments). Based on this, we concluded that linear was the only method that was most accurate as it was off only by less than 1%. Cubic and quadratic had a much greater difference in white pixel percentage and even gave a negative number which doesn't make sense as the actual images did display fibrotic tissue. Thus, for limited image analysis like we did in this project, the linear inteprolation method is the best at predicting fibrotic levels. 

Ethical considerations in this project include to avoid overgeneralizing the results especially since we initially had a limited image analysis. It's important to conduct further testing and try out different scenarios with the image interpolation to ensure that our findings are consistent. It's also important to note that our results were for mouse lung tissue and not human tissue, so that's another important factor to consider in generalizing the results.  There's also the issue of animal safety as the images from this dataset were of mouse lung tissue. It's critical that the mice here were treated in an ethical matter as even that can impact results if animals feel stressed during lab testing. 


## Limitations and Future Work:

The analysis was based on a small number of lung images taken at a limited set of depths. The observed trend between tissue depth and percent white pixels cannot be generalized to all patients with idiopathic pulmonary fibrosis. 
Also, using the percentage of white pixels as a proxy for fibrosis is a simplification. The result depends on image quality, staining consistency, noise, and the threshold chosen to classify pixels as black or white. 
The interpolation step is also limited because it estimates values between a small number of unevenly spaced data points and assumes the relationship changes smoothly with depth. This may not fully capture the complexity of fibrotic tissue.

With more time, we would expand the dataset to include more images, more depth measurements, and samples from multiple patients so the analysis would be more statistically reliable. 
We would also test improved image-processing methods such as adaptive thresholding, better segmentation, or machine-learning-based classification to measure fibrosis more accurately.
